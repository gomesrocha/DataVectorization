## 8. Filtros Complexos com $and   e   $or

```python
# Filtro com $and (TODAS as condições devem ser verdadeiras)
print("FILTRO COM $AND (TODAS as condições)")
print("="*80)

resultados_and = collection.query(
    query_texts=["tecnologia"],
    n_results=10,
    where={
        "$and": [
            {"categoria": {"$eq": "ia"}},
            {"ano": {"$gte": 2024}}
        ]
    }
)

print(f"Documentos de categoria 'ia' E ano >= 2024:")
for i, doc in enumerate(resultados_and['documents'][0]):
    print(f"  {i+1}. {doc[:60]}...")

# Filtro com $or (QUALQUER UMA das condições pode ser verdadeira)
print(f"\n\nFILTRO COM $OR (QUALQUER UMA das condições)")
print("="*80)

resultados_or = collection.query(
    query_texts=["dados"],
    n_results=10,
    where={
        "$or": [
            {"categoria": "banco_de_dados"},
            {"categoria": "ia"}
        ]
    }
)

print(f"Documentos de categoria 'banco_de_dados' OU 'ia':")
for i, (doc, metadata) in enumerate(zip(
    resultados_or['documents'][0],
    resultados_or['metadatas'][0]
)):
    print(f"  {i+1}. [{metadata['categoria']}] {doc[:50]}...")
```

## 15. Exemplo Completo: Sistema de Busca de Documentos

```python
# Criar um sistema de busca simples

def buscar_documentos(colecao, pergunta, categoria=None, n_resultados=3):
    """
    Busca documentos na coleção.

    Args:
        colecao: Coleção do ChromaDB
        pergunta: Texto da pergunta
        categoria: Filtro opcional de categoria
        n_resultados: Número de resultados desejados

    Returns:
        Lista de resultados com documentos, distâncias e metadados
    """

    where_filter = None
    if categoria:
        where_filter = {"categoria": {"$eq": categoria}}

    resultados = colecao.query(
        query_texts=[pergunta],
        n_results=n_resultados,
        where=where_filter
    )

    return resultados

# Usar a função
print("SISTEMA DE BUSCA DE DOCUMENTOS")
print("="*80)

# Busca 1: Sem filtro
print("\n1. Busca geral por 'inteligência artificial'")
resultado = buscar_documentos(collection, "inteligência artificial", n_resultados=2)
for i, doc in enumerate(resultado['documents'][0]):
    print(f"   {i+1}. {doc[:70]}...")

# Busca 2: Com filtro de categoria
print("\n2. Busca por 'dados' apenas na categoria 'ia'")
resultado = buscar_documentos(collection, "dados", categoria="ia", n_resultados=2)
for i, doc in enumerate(resultado['documents'][0]):
    print(f"   {i+1}. {doc[:70]}...")

# Busca 3: Outra busca
print("\n3. Busca por 'programação' apenas na categoria 'programacao'")
resultado = buscar_documentos(collection, "programação", categoria="programacao", n_resultados=2)
for i, doc in enumerate(resultado['documents'][0]):
    print(f"   {i+1}. {doc[:70]}...")
```

```python
def listar_modelos_free_openrouter(max_modelos=20):
    """
    Lista modelos do OpenRouter cujo preço de prompt e completion é 0.
    A disponibilidade pode mudar, então este método evita deixar o notebook preso a um modelo antigo.
    """
    url = "https://openrouter.ai/api/v1/models"
    resp = requests.get(url, timeout=30)
    resp.raise_for_status()
    data = resp.json()["data"]

    modelos_free = []
    for m in data:
        pricing = m.get("pricing", {})
        prompt_price = str(pricing.get("prompt", ""))
        completion_price = str(pricing.get("completion", ""))

        if prompt_price in ("0", "0.0", "0.000000") and completion_price in ("0", "0.0", "0.000000"):
            modelos_free.append({
                "id": m.get("id"),
                "name": m.get("name"),
                "context_length": m.get("context_length"),
                "prompt_price": prompt_price,
                "completion_price": completion_price,
            })

    return modelos_free[:max_modelos]

try:
    modelos_free = listar_modelos_free_openrouter()
    df_modelos_free = pd.DataFrame(modelos_free)
    display(df_modelos_free)

    OPENROUTER_MODEL = os.getenv("OPENROUTER_MODEL") or (modelos_free[0]["id"] if modelos_free else None)
    print("Modelo OpenRouter selecionado:", OPENROUTER_MODEL)
except Exception as e:
    print("Não foi possível listar modelos automaticamente:", e)
    # Ajuste manual se necessário
    OPENROUTER_MODEL = os.getenv("OPENROUTER_MODEL") or "tencent/hy3:free"
    print("Modelo fallback:", OPENROUTER_MODEL)
```

# **Coletar notícias de Cybersegurança**


```python

NEWS_FEEDS = {
    "The Hacker News": "https://feeds.feedburner.com/TheHackersNews",
    "BleepingComputer": "https://www.bleepingcomputer.com/feed/",
    "KrebsOnSecurity": "https://krebsonsecurity.com/feed/",
    "CISA": "https://www.cisa.gov/news.xml",
    "cshub": "https://www.cshub.com/rss/articles",
}

def limpar_texto(texto):
    texto = texto or ""
    texto = re.sub(r"\s+", " ", texto).strip()
    return texto

def fallback_extract_text(url, timeout=20):
    """
    Fallback simples caso o Goose não consiga extrair texto suficiente.
    """
    headers = {
        "User-Agent": "Mozilla/5.0 (compatible; Colab-RAG-CyberNews/1.0)"
    }
    resp = requests.get(url, headers=headers, timeout=timeout)
    resp.raise_for_status()

    soup = BeautifulSoup(resp.text, "html.parser")
    for tag in soup(["script", "style", "noscript"]):
        tag.decompose()

    paragraphs = [limpar_texto(p.get_text(" ")) for p in soup.find_all("p")]
    paragraphs = [p for p in paragraphs if len(p) > 80]
    return limpar_texto(" ".join(paragraphs))

def extrair_artigo_com_goose(url):
    """
    Extrai título, descrição e texto principal de uma notícia usando Goose3.
    Se o texto ficar muito curto, usa fallback com BeautifulSoup.
    """
    g = Goose({
        "browser_user_agent": "Mozilla/5.0 (compatible; Colab-RAG-CyberNews/1.0)",
        "use_meta_language": False,
        "target_language": "en",
    })

    article = g.extract(url=url)
    title = limpar_texto(article.title)
    meta_description = limpar_texto(article.meta_description)
    text = limpar_texto(article.cleaned_text)

    if len(text) < 500:
        try:
            text = fallback_extract_text(url)
        except Exception:
            pass

    return {
        "title": title,
        "description": meta_description,
        "text": text,
        "url": url,
    }

def coletar_noticias(feeds, max_por_fonte=3, pausa_segundos=1.0):
    """
    Lê os feeds RSS, extrai os artigos e devolve uma lista de notícias.
    """
    noticias = []

    for source, feed_url in feeds.items():
        print(f"\nFonte: {source}")
        feed = feedparser.parse(feed_url)
        entries = feed.entries[:max_por_fonte]

        for entry in tqdm(entries, desc=source):
            url = entry.get("link")
            if not url:
                continue

            try:
                artigo = extrair_artigo_com_goose(url)
                texto = artigo["text"]

                if len(texto) < 300:
                    print(f"Texto muito curto, ignorando: {url}")
                    continue

                noticias.append({
                    "source": source,
                    "feed_url": feed_url,
                    "url": url,
                    "title": artigo["title"] or limpar_texto(entry.get("title", "")),
                    "published": limpar_texto(entry.get("published", "")),
                    "description": artigo["description"] or limpar_texto(entry.get("summary", "")),
                    "text": texto,
                })

                time.sleep(pausa_segundos)

            except Exception as e:
                print(f"Erro ao processar {url}: {e}")

    return noticias

noticias = coletar_noticias(NEWS_FEEDS, max_por_fonte=2)

df_noticias = pd.DataFrame(noticias)
print("Total de notícias coletadas:", len(df_noticias))

if len(df_noticias) > 0:
    display(df_noticias[["source", "title", "published", "url"]])
else:
    print("Nenhuma notícia coletada. Teste aumentar max_por_fonte ou trocar os feeds.")
```

QUebrar as noticias em chunks, porém cada chunk deve manter os metadados originais

```python

def chunk_text(texto, chunk_size=900, overlap=150):
    """
    Chunking simples por palavras.
    - chunk_size: quantidade aproximada de palavras por chunk
    - overlap: sobreposição entre chunks
    """
    palavras = texto.split()
    chunks = []

    start = 0
    while start < len(palavras):
        end = min(start + chunk_size, len(palavras))
        chunk = " ".join(palavras[start:end]).strip()

        if chunk:
            chunks.append(chunk)

        if end == len(palavras):
            break

        start = max(0, end - overlap)

    return chunks

def gerar_chunks_noticias(noticias, chunk_size=700, overlap=120):
    documentos = []

    for noticia in noticias:
        chunks = chunk_text(noticia["text"], chunk_size=chunk_size, overlap=overlap)

        for idx, chunk in enumerate(chunks):
            base_id = f'{noticia["url"]}#{idx}'
            doc_id = hashlib.sha1(base_id.encode("utf-8")).hexdigest()

            documentos.append({
                "id": doc_id,
                "document": chunk,
                "metadata": {
                    "source": noticia["source"],
                    "title": noticia["title"],
                    "url": noticia["url"],
                    "published": noticia["published"],
                    "chunk_index": idx,
                }
            })

    return documentos

documentos_rag = gerar_chunks_noticias(noticias)

print("Total de chunks:", len(documentos_rag))
if documentos_rag:
    print("\nExemplo de chunk:")
    print("Título:", documentos_rag[0]["metadata"]["title"])
    print("Fonte:", documentos_rag[0]["metadata"]["source"])
    print("URL:", documentos_rag[0]["metadata"]["url"])
    print(textwrap.shorten(documentos_rag[0]["document"], width=700))
```

Gerar embeddings e indexar no ChromaDB

```python

def embed_texts_gemini(textos, tipo="documento", output_dimensionality=768, pausa_segundos=0.1):
    """
    Gera embeddings com Gemini.
    Para o Gemini Embedding 2, usamos instruções no próprio texto para diferenciar documento e consulta.
    """
    embeddings = []

    for texto in tqdm(textos, desc=f"Gerando embeddings ({tipo})"):
        if tipo == "consulta":
            conteudo = f"Represent this question for retrieving relevant cybersecurity news chunks:\n\n{texto}"
        else:
            conteudo = f"Represent this cybersecurity news chunk for retrieval:\n\n{texto}"

        result = gemini_client.models.embed_content(
            model=GEMINI_EMBEDDING_MODEL,
            contents=conteudo,
            config=types.EmbedContentConfig(output_dimensionality=output_dimensionality),
        )

        embeddings.append(result.embeddings[0].values)
        time.sleep(pausa_segundos)

    return embeddings

# Cliente persistente: os dados ficam em /content enquanto a sessão do Colab existir.
client_cyber = chromadb.PersistentClient(path="/content/chroma_cyber_news")

collection_cyber = client_cyber.get_or_create_collection(
    name="cyber_news_gemini_768",
    metadata={
        "descricao": "RAG com notícias de cibersegurança usando Gemini embeddings",
        "hnsw:space": "cosine",
    }
)

print("Coleção:", collection_cyber.name)
print("Quantidade atual:", collection_cyber.count())
```

```python
# Indexação dos chunks

if not documentos_rag:
    raise ValueError("Não há documentos para indexar. Execute a coleta de notícias primeiro.")

ids = [d["id"] for d in documentos_rag]
docs = [d["document"] for d in documentos_rag]
metas = [d["metadata"] for d in documentos_rag]

embeddings_docs = embed_texts_gemini(docs, tipo="documento", output_dimensionality=768)

collection_cyber.upsert(
    ids=ids,
    documents=docs,
    metadatas=metas,
    embeddings=embeddings_docs,
)

print("Indexação concluída.")
print("Total na coleção:", collection_cyber.count())
```

Recuperação semântica dos dados

```python

def buscar_contextos(pergunta, n_resultados=3, fonte=None):
    query_embedding = embed_texts_gemini(
        [pergunta],
        tipo="consulta",
        output_dimensionality=768,
        pausa_segundos=0
    )[0]

    where_filter = None
    if fonte:
        where_filter = {"source": {"$eq": fonte}}

    resultados = collection_cyber.query(
        query_embeddings=[query_embedding],
        n_results=n_resultados,
        where=where_filter,
        include=["documents", "metadatas", "distances"],
    )

    contextos = []
    for doc, meta, dist in zip(
        resultados["documents"][0],
        resultados["metadatas"][0],
        resultados["distances"][0],
    ):
        contextos.append({
            "text": doc,
            "metadata": meta,
            "distance": dist,
        })

    return contextos

pergunta_teste = "Quais as novas técnicas de ataque voltada a assistentes de codificação de Inteligência Artificial (IA)?"
contextos = buscar_contextos(pergunta_teste, n_resultados=5)

for i, ctx in enumerate(contextos, start=1):
    meta = ctx["metadata"]
    print(f"\n[{i}] distância={ctx['distance']:.4f}")
    print("Fonte:", meta["source"])
    print("Título:", meta["title"])
    print("URL:", meta["url"])
    print(textwrap.shorten(ctx["text"], width=500))
```

Montando o RAG: Prompt contextualizado

```python
def montar_prompt_rag(pergunta, contextos):
    blocos = []

    for i, ctx in enumerate(contextos, start=1):
        meta = ctx["metadata"]
        blocos.append(
            f"""[Fonte {i}]
Título: {meta.get("title")}
Origem: {meta.get("source")}
Data: {meta.get("published")}
URL: {meta.get("url")}
Trecho:
{ctx["text"]}
"""
        )

    contexto_formatado = "\n\n".join(blocos)

    prompt = f"""
Você é um especialista em cybersegurança e avalia noticias sobre o tema.

Responda em português do Brasil.
Use APENAS os contextos fornecidos, porém traduzindo para o português.
Cite as fontes no formato [URL 1], [URL 2] etc.
Se os contextos forem insuficientes, diga claramente que a base não traz evidência suficiente.
Não invente CVEs, datas, empresas, números ou atribuições.
Sempre traga a URL da fonte.

Pergunta:
{pergunta}

Contextos recuperados:
{contexto_formatado}

Resposta:
"""
    return prompt.strip()

prompt = montar_prompt_rag(pergunta_teste, contextos)
print(prompt[:4000])
```

RAG: Geração de respostas usando o OpenRouter

```python
import requests
from google.colab import userdata

OPENROUTER_API_KEY = userdata.get("OPENROUTER_API_KEY")

def limpar_api_key(api_key: str) -> str:
    if not api_key:
        raise ValueError("OPENROUTER_API_KEY não encontrada. Configure a chave nos Secrets do Colab.")

    return api_key.strip().replace("\r", "").replace("\n", "")

def gerar_resposta_openrouter(pergunta, contextos, modelo="cohere/north-mini-code:free"):
    api_key = limpar_api_key(OPENROUTER_API_KEY)

    contexto_formatado = "\n\n".join(
        [
            f"Fonte {i+1}:\n{doc}"
            for i, doc in enumerate(contextos)
        ]
    )

    prompt = f"""
Você é um assistente especializado em cibersegurança.

Responda à pergunta usando apenas o contexto fornecido.
Se a resposta não estiver no contexto, diga que não há informação suficiente.

Contexto:
{contexto_formatado}

Pergunta:
{pergunta}

Resposta:
"""

    url = "https://openrouter.ai/api/v1/chat/completions"

    headers = {
        "Authorization": f"Bearer {api_key}",
        "Content-Type": "application/json",
        "HTTP-Referer": "https://colab.research.google.com",
        "X-Title": "RAG Cybersecurity Colab"
    }

    payload = {
        "model": modelo,
        "messages": [
            {
                "role": "system",
                "content": "Você responde perguntas usando RAG e deve evitar inventar informações. e ao final, sempre traga a URL da noticia"
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        "temperature": 0.2
    }

    response = requests.post(url, headers=headers, json=payload, timeout=60)

    if response.status_code != 200:
        print("Status:", response.status_code)
        print("Resposta:", response.text)
        response.raise_for_status()

    data = response.json()
    return data["choices"][0]["message"]["content"]
```

```python
resposta_openrouter = gerar_resposta_openrouter(pergunta_teste, contextos)
print(resposta_openrouter)
```

Gerando com Gemini

```python
def gerar_resposta_gemini(pergunta, contextos, model=GEMINI_GENERATION_MODEL, temperature=0.2):
    prompt = montar_prompt_rag(pergunta, contextos)

    response = gemini_client.models.generate_content(
        model=model,
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=temperature,
            max_output_tokens=900,
        ),
    )

    return response.text

resposta_gemini = gerar_resposta_gemini(pergunta_teste, contextos)
print(resposta_gemini)
```

```python
def perguntar_rag(
    pergunta,
    provider="openrouter",
    n_resultados=5,
    fonte=None,
    modelo_openrouter="cohere/north-mini-code:free"
):
    contextos = buscar_contextos(
        pergunta,
        n_resultados=n_resultados,
        fonte=fonte
    )

    print("Contextos recuperados:")
    for i, ctx in enumerate(contextos, start=1):
        meta = ctx["metadata"]

        print(
            f"[{i}] {meta.get('source')} | "
            f"{meta.get('title')} | "
            f"distância={ctx['distance']:.4f}"
        )
        print(meta.get("url"))
        print()

    provider = provider.lower().strip()

    if provider == "openrouter":
        resposta = gerar_resposta_openrouter(
            pergunta,
            contextos,
            modelo=modelo_openrouter
        )

    elif provider == "gemini":
        resposta = gerar_resposta_gemini(
            pergunta,
            contextos
        )

    else:
        raise ValueError("provider deve ser 'openrouter' ou 'gemini'.")

    return resposta

pergunta = "Quais incidentes ou vulnerabilidades recentes parecem mais relevantes para uma empresa monitorar?"
#print(perguntar_rag(pergunta, n_resultados=5))
print(perguntar_rag(pergunta, provider="gemini", n_resultados=5))
```
